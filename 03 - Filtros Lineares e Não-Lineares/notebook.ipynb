```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
```

```python
img = cv2.cvtColor(data.coffee(), cv2.COLOR_RGB2GRAY)
plt.imshow(img, cmap='gray')
```

```python
def implot(img, img_filt, nomefiltro, kernelsize):
  plt.subplot(1,2,1)
  plt.imshow(img, cmap='gray')
  plt.title('Imagem original')
  plt.axis('off')

  plt.subplot(1,2,2)
  plt.imshow(img_filt, cmap='gray')
  plt.title('Filtro ' + nomefiltro + ' ' + kernelsize)
  plt.axis('off')

  plt.show()
```

```python
def media(img, kernelsize):
  altura = img.shape[0]
  largura = img.shape[1]

  img_pad = np.pad(img, pad_width=(1,1), mode='reflect')

  res = np.zeros_like(img)

  for i in range(altura):
    for j in range(largura):
      janela = img_pad[i : i + kernelsize, j : j + kernelsize]
      res[i, j] = np.sum(janela) / kernelsize**2

  return res
```

```python
def mediana(img, kernelsize):
  altura = img.shape[0]
  largura = img.shape[1]

  img_pad = np.pad(img, pad_width=(1,1), mode='reflect')

  res = np.zeros_like(img)

  for i in range(altura):
    for j in range(largura):
      janela = img_pad[i : i + kernelsize, j : j + kernelsize]
      res[i, j] = np.median(janela)

  return res
```

```python
def moda(img, kernelsize):
  altura = img.shape[0]
  largura = img.shape[1]

  img_pad = np.pad(img, pad_width=(1,1), mode='reflect')

  res = np.zeros_like(img)

  for i in range(altura):
    for j in range(largura):
      janela = img_pad[i : i + kernelsize, j : j + kernelsize]
      janela_flat = janela.flatten()
      res[i, j] = np.bincount(janela_flat).argmax()

  return res
```

```python
def maxima(img, kernelsize):
  altura = img.shape[0]
  largura = img.shape[1]

  img_pad = np.pad(img, pad_width=(1,1), mode='reflect')

  res = np.zeros_like(img)

  for i in range(altura):
    for j in range(largura):
      janela = img_pad[i : i + kernelsize, j : j + kernelsize]
      janela_flat = janela.flatten()
      res[i, j] = max(janela_flat)

  return res
```

```python
img_media = media(img, 5)
implot(img, img_media, 'média', '5x5')
```

```python
img_mediana = mediana(img, 5)
implot(img, img_mediana, 'mediana', '5x5')
```

```python
img_moda = moda(img, 5)
implot(img, img_moda, 'moda', '5x5')
```

```python
img_maxima = maxima(img, 5)
implot(img, img_maxima, 'máxima', '5x5')
```
