```python
img_bgr = cv2.imread('sssdfg.png')
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
```

```python
altura = img_gray.shape[0]
largura = img_gray.shape[1]

print(altura, largura)
```

```python
dx = 100
dy = 300

transladada = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    novo_x = x + dx
    novo_y = y + dy

    if 0 <= novo_x and novo_x < largura and 0 <= novo_y and novo_y < altura:
      transladada[novo_y, novo_x] = img_gray[y, x]

plt.figure(figsize=(7,5))
plt.imshow(transladada, cmap='gray')
plt.show()
```

```python
dx = 100
dy = 300

wraparound = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    novo_x = (x + dx) % largura
    novo_y = (y + dy) % altura

    wraparound[novo_y, novo_x] = img_gray[y, x]

plt.figure(figsize=(7,5))
plt.imshow(wraparound, cmap='gray')
plt.show()
```

```python
sx = 2
sy = 0.5

nova_largura = int(largura * sx)
nova_altura = int(altura * sy)

escalada = np.zeros((nova_altura, nova_largura), dtype=np.uint8)

for x in range(nova_largura):
  for y in range(nova_altura):
    novo_x = int(x/sx)
    novo_y = int(y/sy)

    escalada[y, x] = img_gray[novo_y, novo_x]

plt.figure(figsize=(7,5))
plt.imshow(escalada, cmap='gray')
plt.show()
```

```python
espelhada_x = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    novo_x = largura - 1 - x
    novo_y = y

    espelhada_x[y, x] = img_gray[novo_y, novo_x]

plt.figure(figsize=(7,5))
plt.imshow(espelhada_x, cmap='gray')
plt.show()
```

```python
espelhada_y = np.zeros((altura, largura), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    novo_x = x
    novo_y = altura - 1 - y

    espelhada_y[y, x] = img_gray[novo_y, novo_x]

plt.figure(figsize=(7,5))
plt.imshow(espelhada_y, cmap='gray')
plt.show()
```
